```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Iris.csv")
df
```

```python
print(df.shape)
print(df.columns)
```

```python
df.describe(include="all")
```

```python
df.info()
```

```python
df.isnull().sum()
df = df.dropna()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
sns.set_style("whitegrid")
```

```python
plt.figure(figsize={8,5})
sns.histplot(data=df, x="SepalLengthCm", bins=10)
plt.title("Distribution of Sepia length")
plt.xlabel("Sepal length(cm)")
plt.ylabel("Number of flowers")
```

```python
plt.figure(figsize={8,5})
sns.boxplot(data=df, x="Species", y="SepalLengthCm")
plt.title("Sepal lenght by species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
```

```python
plt.figure(figsize={8,5})
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
plt.title("Sepal Len vs Petal len")
plt.xlabel("Sepal Length (Cm)")
```

```python
sns.pairplot(df, hue="Species")
```

```python
numeric_columns = ['Id', 'SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
correlation = df[numeric_columns].corr()
correlation
```

```python
plt.figure(figsize={8,5})
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
```
