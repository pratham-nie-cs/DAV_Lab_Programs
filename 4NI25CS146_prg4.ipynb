```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("melb_data.csv")
print(df.head())
print(df.shape)
print(df.columns)
```

```python
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
df_mean = df.copy()
mean_building_area = df_mean['BuildingArea'].mean()
```

```python
df_mean['BuildingArea'] = df_mean['BuildingArea'].fillna(mean_building_area)
df_mean['BuildingArea'].isnull().sum()
```

```python
df_median = df.copy()
median_year_built = df['YearBuilt'].median()
df_median['YearBuilt'] = df_median['YearBuilt'].fillna(median_year_built)
median_year_built
```

```python
df['YearBuilt'].isnull().sum()
```

```python
df_ffill = df.copy()
df_ffill['Car'] = df_ffill['Car'].ffill()
```

```python
df_ffill['Car'].isnull().sum()
```
